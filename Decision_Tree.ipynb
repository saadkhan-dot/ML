```python
# 2. Create dataset

data = {
    "Customer" : [
        "C1", "C2", "C3", "C4", "C5", "C6", "C7", "C8", "C9", "C10"],
    "Income" : [
        25, 30, 35, 40, 45, 50, 55, 60, 65, 70],
    "CreditScore" : [
        600, 620, 650, 680, 700, 710, 720, 730, 740, 750],
    "Loan":[
        "No", "No", "No", "No", "Yes", "Yes", "Yes", "Yes", "Yes", "Yes"]
}
df = pd.DataFrame(data)
```

```python
print("\n"+ "=" * 60)
print("Loan Distribution")
print("="*60)
print(df["Loan"])
```

```python
print("\n" + "=" * 60)
print("Root Node - Gini Impurity")
print("="*60)

total = len(df)

yes_count = (df["Loan"]=="Yes").sum()
no_count = (df["Loan"]=="No").sum()

p_yes = yes_count / total
p_no = no_count / total

gini_root = 1 - (p_yes ** 2 + p_no ** 2)

print("Total Customers:", total)
print("Yes Count:", yes_count)
print("No Count:", no_count)

print("\nP(yes) =", p_yes)
print("P(no) =", p_no)

print("\nGini Impurity of Root Node:", round(gini_root, 4))
```

```python
print("\nP(Yes)", p_yes)
print("P(No)", p_no)

print("\nGini Impurity:")
print("Gini = 1 - (P(yes)^2 + P(no)^2)")
print("Gini = 1 - (", p_yes, "^2 +", p_no, "^2)")
print(
    f"Gini = 1 -({p_yes:.2f}^2 + {p_no:.2f}^2)"
)
print("Root Gini =", round(gini_root, 4))
```
